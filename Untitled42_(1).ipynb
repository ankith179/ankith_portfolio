````python
# Cell 1.10: Detect the number of patients available in the actual TCGA and METABRIC clinical datasets
# This cell does NOT use patient counts reported in the paper.

import pandas as pd

def count_clinical_patients(clinical_file, skiprows=4):
    """
    Load a cBioPortal clinical patient file and return:
    - number of patients
    - dataframe
    """

    clinical_df = pd.read_csv(
        clinical_file,
        sep="\t",
        skiprows=skiprows,
        low_memory=False
    )

    # Remove completely empty rows
    clinical_df = clinical_df.dropna(how="all")

    # Identify the patient ID column
    possible_id_columns = [
        "PATIENT_ID",
        "Patient Identifier",
        "PATIENT",
        "PATIENTID"
    ]

    patient_id_column = next(
        (col for col in possible_id_columns if col in clinical_df.columns),
        None
    )

    if patient_id_column is None:
        raise ValueError(
            f"Patient ID column not found in {clinical_file.name}. "
            f"Available columns: {clinical_df.columns.tolist()[:20]}"
        )

    # Count unique non-null patient IDs
    patient_count = (
        clinical_df[patient_id_column]
        .dropna()
        .astype(str)
        .str.strip()
        .replace("", pd.NA)
        .dropna()
        .nunique()
    )

    return patient_count, clinical_df


# Actual clinical files
TCGA_CLINICAL_FILE = TCGA_DIR / "data_clinical_patient.txt"
METABRIC_CLINICAL_FILE = METABRIC_DIR / "data_clinical_patient.txt"


# Detect actual patient counts from the files
TCGA_PATIENT_COUNT, tcga_clinical_preview = count_clinical_patients(
    TCGA_CLINICAL_FILE
)

METABRIC_PATIENT_COUNT, metabric_clinical_preview = count_clinical_patients(
    METABRIC_CLINICAL_FILE
)

COMBINED_PATIENT_COUNT = (
    TCGA_PATIENT_COUNT +
    METABRIC_PATIENT_COUNT
)


# Store configuration based on the actual data
UGES_CONFIG = {
    "random_seed": RANDOM_SEED,
    "tcga_patients_available": TCGA_PATIENT_COUNT,
    "metabric_patients_available": METABRIC_PATIENT_COUNT,
    "combined_patients_available": COMBINED_PATIENT_COUNT,
    "missingness_threshold": 0.50,
    "knn_neighbors": 10,
    "subtypes": [
        "Basal",
        "Her2",
        "Luminal A",
        "Luminal B"
    ],
}


print("================ UGES CONFIGURATION ================")

for key, value in UGES_CONFIG.items():
    print(f"{key}: {value}")
````

````python
# ============================================================
# Cell 5.4: Remove Missing Clinical Labels
# This cell removes patients for whom the subtype label is
# unavailable or represented by a missing-value string.
# ============================================================

missing_subtype_values = [
    "",
    "NA",
    "N/A",
    "NaN",
    "nan",
    "None",
    "NONE",
    "NULL"
]


# ------------------------------------------------------------
# Clean patient IDs
# ------------------------------------------------------------

tcga_labels["PATIENT_ID"] = (
    tcga_labels["PATIENT_ID"]
    .astype("string")
    .str.strip()
)

metabric_labels["PATIENT_ID"] = (
    metabric_labels["PATIENT_ID"]
    .astype("string")
    .str.strip()
)


# ------------------------------------------------------------
# Clean subtype values
# ------------------------------------------------------------

tcga_labels["SUBTYPE"] = (
    tcga_labels["SUBTYPE"]
    .astype("string")
    .str.strip()
)

metabric_labels["SUBTYPE"] = (
    metabric_labels["SUBTYPE"]
    .astype("string")
    .str.strip()
)


# Convert textual missing values to actual NA
tcga_labels["SUBTYPE"] = (
    tcga_labels["SUBTYPE"]
    .replace(missing_subtype_values, pd.NA)
)

metabric_labels["SUBTYPE"] = (
    metabric_labels["SUBTYPE"]
    .replace(missing_subtype_values, pd.NA)
)


# ------------------------------------------------------------
# Remove rows without patient IDs or subtype labels
# ------------------------------------------------------------

tcga_labels = (
    tcga_labels
    .dropna(
        subset=[
            "PATIENT_ID",
            "SUBTYPE"
        ]
    )
    .copy()
)

metabric_labels = (
    metabric_labels
    .dropna(
        subset=[
            "PATIENT_ID",
            "SUBTYPE"
        ]
    )
    .copy()
)


# ------------------------------------------------------------
# Remove duplicate patient IDs
# ------------------------------------------------------------

tcga_duplicate_patients = (
    tcga_labels["PATIENT_ID"]
    .duplicated()
    .sum()
)

metabric_duplicate_patients = (
    metabric_labels["PATIENT_ID"]
    .duplicated()
    .sum()
)


print("=" * 80)
print("CLINICAL LABEL CLEANING")
print("=" * 80)

print(
    "TCGA duplicate patient IDs:",
    tcga_duplicate_patients
)

print(
    "METABRIC duplicate patient IDs:",
    metabric_duplicate_patients
)


tcga_labels = (
    tcga_labels
    .drop_duplicates(
        subset="PATIENT_ID",
        keep="first"
    )
    .reset_index(drop=True)
)

metabric_labels = (
    metabric_labels
    .drop_duplicates(
        subset="PATIENT_ID",
        keep="first"
    )
    .reset_index(drop=True)
)


print("\nAfter cleaning:")

print(
    "TCGA patients:",
    len(tcga_labels)
)

print(
    "METABRIC patients:",
    len(metabric_labels)
)
````

````python
import matplotlib.pyplot as plt
import seaborn as sns

# Define the subtypes of interest
subtypes_of_interest = ['Basal', 'Her2', 'LumA', 'LumB']

# Re-applying subtype standardization and filtering to tcga_clinical_std and metabric_clinical_std.
# This logic ensures the 'Subtype' column exists and contains cleaned, UGES-eligible labels.

# Process TCGA subtypes: clean names and handle missing values
tcga_clinical_std["Subtype"] = (
    tcga_clinical_std["SUBTYPE"]
    .astype(str)
    .str.strip()
    .str.replace("BRCA_", "", regex=False)
    .replace(missing_subtype_values, pd.NA)
)

# Process METABRIC subtypes: clean names and handle missing values
metabric_clinical_std["Subtype"] = (
    metabric_clinical_std["CLAUDIN_SUBTYPE"]
    .astype(str)
    .str.strip()
    .replace({"claudin-low": "Claudin-low"}, regex=False) # Specific replacement for claudin-low
    .replace(missing_subtype_values, pd.NA)
)

# Filter both dataframes to keep only the UGES-eligible subtypes and drop rows with missing 'Subtype'
tcga_clinical_std = tcga_clinical_std[tcga_clinical_std['Subtype'].isin(subtypes_of_interest)].dropna(subset=['Subtype']).copy()
metabric_clinical_std = metabric_clinical_std[metabric_clinical_std['Subtype'].isin(subtypes_of_interest)].dropna(subset=['Subtype']).copy()

# Filter and count subtypes for TCGA
tcga_subtypes_counts = (
    tcga_clinical_std[tcga_clinical_std['Subtype'].isin(subtypes_of_interest)]
    ['Subtype']
    .value_counts()
    .rename_axis('Subtype')
    .reset_index(name='Count')
)
tcga_subtypes_counts['Dataset'] = 'TCGA'

# Filter and count subtypes for METABRIC
metabric_subtypes_counts = (
    metabric_clinical_std[metabric_clinical_std['Subtype'].isin(subtypes_of_interest)]
    ['Subtype']
    .value_counts()
    .rename_axis('Subtype')
    .reset_index(name='Count')
)
metabric_subtypes_counts['Dataset'] = 'METABRIC'

# Combine the counts into a single DataFrame
combined_subtype_counts = pd.concat([tcga_subtypes_counts, metabric_subtypes_counts])

# Set up the plot style
sns.set_theme(style="whitegrid")
plt.figure(figsize=(8, 4))

# Create the bar plot
ax = sns.barplot(data=combined_subtype_counts, x='Subtype', y='Count', hue='Dataset', palette='viridis')

# Add title and labels
plt.title('Distribution of Molecular Subtypes in TCGA and METABRIC Cohorts')
plt.xlabel('Molecular Subtype')
plt.ylabel('Number of Patients')
plt.legend(title='Dataset')
plt.tight_layout()

# Add numbers on top of the bars
for container in ax.containers:
    ax.bar_label(container, fmt='%d', label_type='edge', padding=3)

plt.show()
````

````python
# ============================================================
# Cell 5.5: Standardize Molecular Subtype Names
# This cell converts different subtype naming conventions into
# one consistent set of labels for both datasets.
# ============================================================

def standardize_subtype(value):
    """
    Convert known subtype naming variants to the common
    UGES naming convention.
    """

    if pd.isna(value):
        return pd.NA

    value = str(value).strip()

    subtype_map = {
        # TCGA labels
        "BRCA_Basal": "Basal",
        "BRCA_Her2": "Her2",
        "BRCA_HER2": "Her2",
        "BRCA_LumA": "LumA",
        "BRCA_LumB": "LumB",
        "BRCA_Normal": "Normal",

        # Common naming variants
        "Basal-like": "Basal",
        "Basal Like": "Basal",

        "HER2": "Her2",
        "HER2-enriched": "Her2",
        "HER2 enriched": "Her2",

        "Luminal A": "LumA",
        "Luminal-A": "LumA",

        "Luminal B": "LumB",
        "Luminal-B": "LumB",

        "claudin-low": "Claudin-low",
        "Claudin-Low": "Claudin-low"
    }

    return subtype_map.get(value, value)


# ------------------------------------------------------------
# Apply standardization
# ------------------------------------------------------------

tcga_labels["Subtype"] = (
    tcga_labels["SUBTYPE"]
    .map(standardize_subtype)
)

metabric_labels["Subtype"] = (
    metabric_labels["SUBTYPE"]
    .map(standardize_subtype)
)


# ------------------------------------------------------------
# Display standardized distributions
# ------------------------------------------------------------

print("=" * 80)
print("STANDARDIZED TCGA SUBTYPES")
print("=" * 80)

display(
    tcga_labels["Subtype"]
    .value_counts(dropna=False)
)


print("\n" + "=" * 80)
print("STANDARDIZED METABRIC SUBTYPES")
print("=" * 80)

display(
    metabric_labels["Subtype"]
    .value_counts(dropna=False)
)
````

````python
# ============================================================
# Cell 5.6: Select UGES Intrinsic Subtypes
# This cell retains only the four intrinsic breast cancer
# subtypes used as the UGES classification targets.
# ============================================================

UGES_SUBTYPES = [
    "Basal",
    "Her2",
    "LumA",
    "LumB"
]


# ------------------------------------------------------------
# Keep only the four UGES classes
# ------------------------------------------------------------

tcga_labels = (
    tcga_labels[
        tcga_labels["Subtype"]
        .isin(UGES_SUBTYPES)
    ]
    .copy()
    .reset_index(drop=True)
)

metabric_labels = (
    metabric_labels[
        metabric_labels["Subtype"]
        .isin(UGES_SUBTYPES)
    ]
    .copy()
    .reset_index(drop=True)
)


# ------------------------------------------------------------
# Display final subtype distributions
# ------------------------------------------------------------

print("=" * 80)
print("TCGA — FINAL UGES SUBTYPE DISTRIBUTION")
print("=" * 80)

display(
    tcga_labels["Subtype"]
    .value_counts()
    .reindex(
        UGES_SUBTYPES,
        fill_value=0
    )
)


print("\n" + "=" * 80)
print("METABRIC — FINAL UGES SUBTYPE DISTRIBUTION")
print("=" * 80)

display(
    metabric_labels["Subtype"]
    .value_counts()
    .reindex(
        UGES_SUBTYPES,
        fill_value=0
    )
)
````

Module 6 — TCGA Mutation Preprocessing

````python
# Cell 6.1: Initialize Module 6 for TCGA mutation preprocessing according to the UGES methodology

print("=" * 80)
print("MODULE 6 : TCGA Mutation Preprocessing")
print("=" * 80)

print("Method described in the paper:")
print("1. Map mutation profiles to gene level")
print("2. Create a binary gene-by-patient matrix")
print("3. Value = 1 if gene is mutated at least once")
print("4. Value = 0 otherwise")
````

````python
# Cell 6.2: Identify the gene and patient/sample identifier columns available in the raw TCGA mutation dataset

print("================ TCGA MUTATION COLUMNS ================")

mutation_columns = tcga_mutation.columns.tolist()

for i, column in enumerate(mutation_columns, start=1):
    print(f"{i:3d}. {column}")
````

````python
# Cell 6.3: Identify the gene symbol column required for gene-level mutation aggregation

possible_gene_columns = [
    "Hugo_Symbol",
    "SYMBOL",
    "Gene",
    "gene",
    "Gene_Symbol"
]

mutation_gene_column = next(
    (
        column
        for column in possible_gene_columns
        if column in tcga_mutation.columns
    ),
    None
)

if mutation_gene_column is None:
    raise KeyError(
        "Could not identify the gene-symbol column in TCGA mutation data."
    )

print("================ MUTATION GENE COLUMN ================")
print(f"Gene column: {mutation_gene_column}")
````

````python
# Cell 6.4: Identify the sample or patient identifier field in the TCGA mutation dataset

possible_sample_columns = [
    "Tumor_Sample_Barcode",
    "Sample_ID",
    "SAMPLE_ID",
    "PATIENT_ID",
    "Patient_ID",
    "patient_id"
]

mutation_sample_column = next(
    (
        column
        for column in possible_sample_columns
        if column in tcga_mutation.columns
    ),
    None
)

if mutation_sample_column is None:
    raise KeyError(
        "Could not identify the TCGA mutation sample identifier column."
    )

print("================ MUTATION SAMPLE COLUMN ================")
print(f"Sample column: {mutation_sample_column}")

print("\nExample identifiers:")
display(
    tcga_mutation[mutation_sample_column]
    .dropna()
    .head(10)
)
````

````python
# Cell 6.5: Inspect gene symbols and mutation records before converting them into the binary gene-by-patient matrix

print("================ MUTATION GENE INSPECTION ================")

print(
    f"Total mutation records : {len(tcga_mutation):,}"
)

print(
    f"Unique genes           : "
    f"{tcga_mutation[mutation_gene_column].nunique():,}"
)

print(
    f"Unique samples         : "
    f"{tcga_mutation[mutation_sample_column].nunique():,}"
)

display(
    tcga_mutation[
        [
            mutation_gene_column,
            mutation_sample_column
        ]
    ].head(10)
)
````

````python
# Cell 6.6: Remove mutation records that lack either a gene symbol or sample identifier before aggregation

tcga_mutation_clean = tcga_mutation[
    tcga_mutation[mutation_gene_column].notna()
    &
    tcga_mutation[mutation_sample_column].notna()
].copy()

print("================ CLEAN MUTATION RECORDS ================")

print(
    f"Original records : {len(tcga_mutation):,}"
)

print(
    f"Valid records    : {len(tcga_mutation_clean):,}"
)

print(
    f"Removed records  : "
    f"{len(tcga_mutation) - len(tcga_mutation_clean):,}"
)
````

````python
# Cell 6.7: Standardize gene symbols before creating the gene-level mutation matrix

tcga_mutation_clean[mutation_gene_column] = (
    tcga_mutation_clean[mutation_gene_column]
    .astype(str)
    .str.strip()
)

tcga_mutation_clean = tcga_mutation_clean[
    tcga_mutation_clean[mutation_gene_column] != ""
].copy()

print("================ GENE SYMBOL STANDARDIZATION ================")

print(
    f"Unique genes after cleaning: "
    f"{tcga_mutation_clean[mutation_gene_column].nunique():,}"
)

display(
    tcga_mutation_clean[
        [mutation_gene_column]
    ]
    .drop_duplicates()
    .head(20)
)
````

````python
# Cell 6.8: Convert TCGA sample barcodes into patient identifiers using the first three barcode components

def tcga_sample_to_patient(sample_id):
    """
    Convert a TCGA sample barcode such as TCGA-3C-AAAU-01
    into the patient identifier TCGA-3C-AAAU.
    """
    if pd.isna(sample_id):
        return np.nan

    sample_id = str(sample_id).strip()

    parts = sample_id.split("-")

    if len(parts) >= 3:
        return "-".join(parts[:3])

    return sample_id


tcga_mutation_clean["PATIENT_ID"] = (
    tcga_mutation_clean[mutation_sample_column]
    .apply(tcga_sample_to_patient)
)

print("================ TCGA PATIENT ID CONVERSION ================")

display(
    tcga_mutation_clean[
        [
            mutation_sample_column,
            "PATIENT_ID"
        ]
    ]
    .drop_duplicates()
    .head(15)
)
````

````python
# Cell 6.9: Check how many molecular samples correspond to each TCGA patient before gene-level aggregation

sample_patient_counts = (
    tcga_mutation_clean[
        [mutation_sample_column, "PATIENT_ID"]
    ]
    .drop_duplicates()
    .groupby("PATIENT_ID")
    .size()
    .sort_values(ascending=False)
)

print("================ TCGA SAMPLE/PATIENT RELATIONSHIP ================")

print(
    f"Unique mutation samples : "
    f"{tcga_mutation_clean[mutation_sample_column].nunique():,}"
)

print(
    f"Unique mutation patients : "
    f"{tcga_mutation_clean['PATIENT_ID'].nunique():,}"
)

print("\nPatients with multiple mutation samples:")

display(
    sample_patient_counts[
        sample_patient_counts > 1
    ].head(20)
)
````

````python
# Cell 6.10: Restrict mutation records to patients present in the standardized TCGA clinical dataset

tcga_clinical_patient_ids = set(
    tcga_clinical_master["PATIENT_ID"]
    .dropna()
    .astype(str)
)

tcga_mutation_clean["PATIENT_ID"] = (
    tcga_mutation_clean["PATIENT_ID"]
    .astype(str)
)

tcga_mutation_linked = tcga_mutation_clean[
    tcga_mutation_clean["PATIENT_ID"].isin(
        tcga_clinical_patient_ids
    )
].copy()

print("================ CLINICAL-MUTATION LINK ================")

print(
    f"Mutation records before linking : "
    f"{len(tcga_mutation_clean):,}"
)

print(
    f"Mutation records after linking  : "
    f"{len(tcga_mutation_linked):,}"
)

print(
    f"Linked patients : "
    f"{tcga_mutation_linked['PATIENT_ID'].nunique():,}"
)
````

````python
# Cell 6.11: Collapse multiple mutation events affecting the same gene in the same patient into one gene-patient event

tcga_gene_patient_events = (
    tcga_mutation_linked[
        [
            mutation_gene_column,
            "PATIENT_ID"
        ]
    ]
    .drop_duplicates()
    .rename(
        columns={
            mutation_gene_column: "Gene"
        }
    )
)

print("================ UNIQUE GENE-PATIENT EVENTS ================")

print(
    f"Unique gene-patient events : "
    f"{len(tcga_gene_patient_events):,}"
)

print(
    f"Unique genes : "
    f"{tcga_gene_patient_events['Gene'].nunique():,}"
)

print(
    f"Unique patients : "
    f"{tcga_gene_patient_events['PATIENT_ID'].nunique():,}"
)

display(
    tcga_gene_patient_events.head(10)
)
````

````python
# Cell 6.12: Construct the binary gene-by-patient mutation matrix required by the UGES methodology

tcga_mutation_binary = pd.crosstab(
    tcga_gene_patient_events["Gene"],
    tcga_gene_patient_events["PATIENT_ID"]
)

# Convert counts to binary indicators.
# A value of 1 means the gene was mutated at least once in that patient.
tcga_mutation_binary = (
    tcga_mutation_binary
    .clip(upper=1)
    .astype(np.int8)
)

print("================ TCGA BINARY MUTATION MATRIX ================")

print(
    f"Rows    (genes)    : {tcga_mutation_binary.shape[0]:,}"
)

print(
    f"Columns (patients) : {tcga_mutation_binary.shape[1]:,}"
)

print(
    f"Matrix values       : "
    f"{sorted(tcga_mutation_binary.stack().unique().tolist())}"
)

display(
    tcga_mutation_binary.iloc[:10, :10]
)
````

````python
# Cell 6.13: Verify that the TCGA mutation matrix contains only valid binary 0/1 values

unique_mutation_values = np.unique(
    tcga_mutation_binary.to_numpy()
)

print("================ BINARY MATRIX VALIDATION ================")

print(
    "Unique values:",
    unique_mutation_values
)

assert set(unique_mutation_values).issubset({0, 1})

print("\n✓ Mutation matrix is strictly binary")
````

````python
# Cell 6.14: Calculate mutation frequencies across patients to inspect the resulting gene-level mutation matrix

tcga_mutation_frequency = pd.DataFrame({
    "Gene": tcga_mutation_binary.index,
    "Mutated_Patients": tcga_mutation_binary.sum(axis=1).values
})

tcga_mutation_frequency["Mutation_Frequency"] = (
    tcga_mutation_frequency["Mutated_Patients"]
    /
    tcga_mutation_binary.shape[1]
)

tcga_mutation_frequency = (
    tcga_mutation_frequency
    .sort_values(
        "Mutated_Patients",
        ascending=False
    )
)

print("================ TCGA MUTATION FREQUENCY ================")

display(
    tcga_mutation_frequency.head(20)
)
````

````python
# Cell 6.15: Inspect several well-known breast-cancer genes as a sanity check of the gene-level mutation matrix

genes_to_check = [
    "TP53",
    "PIK3CA",
    "GATA3",
    "CDH1",
    "ERBB2",
    "AKT1"
]

mutation_sanity_check = tcga_mutation_frequency[
    tcga_mutation_frequency["Gene"].isin(genes_to_check)
].copy()

print("================ MUTATION SANITY CHECK ================")

display(mutation_sanity_check)
````

````python
# ============================================================
# Cell 8.1 : Validate Raw CNA Data
# ============================================================

print("\n" + "=" * 80)
print("CELL 8.1 : RAW CNA DATA VALIDATION")
print("=" * 80)

print("\nTCGA CNA shape:")
print(tcga_cna.shape)

print("\nMETABRIC CNA shape:")
print(metabric_cna.shape)

print("\nTCGA CNA columns:")
print(tcga_cna.columns.tolist()[:20])

print("\nMETABRIC CNA columns:")
print(metabric_cna.columns.tolist()[:20])


if "Hugo_Symbol" not in tcga_cna.columns:
    raise ValueError(
        "TCGA CNA: 'Hugo_Symbol' column not found."
    )

if "Hugo_Symbol" not in metabric_cna.columns:
    raise ValueError(
        "METABRIC CNA: 'Hugo_Symbol' column not found."
    )

print("\n✓ TCGA CNA contains Hugo_Symbol")
print("✓ METABRIC CNA contains Hugo_Symbol")
````

````python
# ============================================================
# Cell 8.2 : Process CNA Matrix
# ============================================================

def process_cna_matrix(df, dataset_name):

    work = df.copy()

    # --------------------------------------------------------
    # Clean gene identifiers
    # --------------------------------------------------------

    work["Hugo_Symbol"] = (
        work["Hugo_Symbol"]
        .astype("string")
        .str.strip()
        .str.upper()
    )

    # Remove invalid gene identifiers
    invalid_genes = {
        "",
        "NAN",
        "NA",
        "NONE",
        "NULL"
    }

    work = work[
        work["Hugo_Symbol"].notna()
        &
        ~work["Hugo_Symbol"].isin(invalid_genes)
    ].copy()

    # --------------------------------------------------------
    # Remove duplicate genes
    # --------------------------------------------------------

    duplicate_genes = (
        work["Hugo_Symbol"]
        .duplicated()
        .sum()
    )

    print(
        f"\n{dataset_name}: duplicate gene rows = "
        f"{duplicate_genes}"
    )

    work = (
        work
        .drop_duplicates(
            subset="Hugo_Symbol",
            keep="first"
        )
        .copy()
    )

    # --------------------------------------------------------
    # Set gene as index
    # --------------------------------------------------------

    work = work.set_index("Hugo_Symbol")

    # --------------------------------------------------------
    # Remove metadata columns if present
    # --------------------------------------------------------

    metadata_columns = [
        "Entrez_Gene_Id",
        "Entrez_Gene_ID",
        "Entrez_Gene_Id_",
        "Gene_ID",
        "Gene_Id"
    ]

    columns_to_remove = [
        col
        for col in metadata_columns
        if col in work.columns
    ]

    if columns_to_remove:
        work = work.drop(
            columns=columns_to_remove
        )

    # --------------------------------------------------------
    # Convert CNA measurements to numeric
    # --------------------------------------------------------

    work = work.apply(
        pd.to_numeric,
        errors="coerce"
    )

    # Remove columns containing no numeric CNA data
    work = work.dropna(
        axis=1,
        how="all"
    )

    # Sort genes for reproducibility
    work = work.sort_index()

    # Final duplicate-index protection
    work = work[
        ~work.index.duplicated(
            keep="first"
        )
    ]

    if work.shape[0] == 0:
        raise ValueError(
            f"{dataset_name}: no CNA genes remain."
        )

    if work.shape[1] == 0:
        raise ValueError(
            f"{dataset_name}: no CNA samples remain."
        )

    print(
        f"{dataset_name} CNA matrix:"
    )

    print(
        "Genes   :",
        work.shape[0]
    )

    print(
        "Samples :",
        work.shape[1]
    )

    return work
````

````python
# ============================================================
# Cell 9.3 : Process TCGA Methylation
# CpG Probe → Gene → Mean Methylation
# ============================================================

def process_tcga_methylation(df):

    df = df.copy()

    required_columns = [
        "ENTITY_STABLE_ID",
        "NAME"
    ]

    missing = [
        col
        for col in required_columns
        if col not in df.columns
    ]

    if missing:
        raise ValueError(
            f"TCGA methylation missing columns: {missing}"
        )

    # --------------------------------------------------------
    # Clean gene mapping
    # --------------------------------------------------------

    df["NAME"] = (
        df["NAME"]
        .astype("string")
        .str.strip()
    )

    # Remove rows without a gene mapping
    df = df[
        df["NAME"].notna()
        &
        (df["NAME"] != "")
        &
        (~df["NAME"].isin([
            "NA",
            "NAN",
            "NONE",
            "NULL"
        ]))
    ].copy()

    # --------------------------------------------------------
    # Identify patient columns
    # --------------------------------------------------------

    annotation_columns = [
        "ENTITY_STABLE_ID",
        "NAME",
        "DESCRIPTION",
        "TRANSCRIPT_ID"
    ]

    patient_columns = [
        col
        for col in df.columns
        if col not in annotation_columns
    ]

    if len(patient_columns) == 0:
        raise ValueError(
            "No TCGA methylation patient columns found."
        )

    # --------------------------------------------------------
    # Convert patient values to numeric
    # --------------------------------------------------------

    df[patient_columns] = (
        df[patient_columns]
        .apply(
            pd.to_numeric,
            errors="coerce"
        )
    )

    # --------------------------------------------------------
    # Some NAME values contain multiple gene symbols.
    #
    # Example:
    # KIAA1409;COX8C
    #
    # Expand these mappings so that each gene receives the
    # corresponding probe measurement.
    # --------------------------------------------------------

    df["NAME"] = (
        df["NAME"]
        .str.split(";")
    )

    df = df.explode(
        "NAME"
    )

    df["NAME"] = (
        df["NAME"]
        .astype("string")
        .str.strip()
        .str.upper()
    )

    # Remove invalid expanded gene names
    df = df[
        df["NAME"].notna()
        &
        (df["NAME"] != "")
    ].copy()

    # --------------------------------------------------------
    # Aggregate CpG probes belonging to the same gene
    #
    # Paper:
    # Multiple CpGs → average methylation for the gene
    # --------------------------------------------------------

    gene_methylation = (
        df
        .groupby(
            "NAME",
            sort=True
        )[patient_columns]
        .mean()
    )

    # Rename index
    gene_methylation.index.name = "Hugo_Symbol"

    # Sort patients for reproducibility
    gene_methylation = (
        gene_methylation
        .reindex(
            sorted(
                gene_methylation.columns
            ),
            axis=1
        )
    )

    return gene_methylation
````

````python
# ============================================================
# Cell 9.5 : Process METABRIC Methylation
# Already Gene-Level
# ============================================================

def process_metabric_methylation(df):

    df = df.copy()

    if "Hugo_Symbol" not in df.columns:
        raise ValueError(
            "METABRIC methylation does not contain "
            "'Hugo_Symbol'."
        )

    # --------------------------------------------------------
    # Clean gene names
    # --------------------------------------------------------

    df["Hugo_Symbol"] = (
        df["Hugo_Symbol"]
        .astype("string")
        .str.strip()
        .str.upper()
    )

    # Remove invalid genes
    df = df[
        df["Hugo_Symbol"].notna()
        &
        (df["Hugo_Symbol"] != "")
        &
        (~df["Hugo_Symbol"].isin([
            "NA",
            "NAN",
            "NONE",
            "NULL"
        ]))
    ].copy()

    # --------------------------------------------------------
    # Remove duplicate gene rows
    #
    # METABRIC is already gene-level.
    # --------------------------------------------------------

    duplicate_genes = (
        df["Hugo_Symbol"]
        .duplicated()
        .sum()
    )

    print(
        "METABRIC duplicate gene rows:",
        duplicate_genes
    )

    df = (
        df
        .drop_duplicates(
            subset="Hugo_Symbol",
            keep="first"
        )
    )

    # --------------------------------------------------------
    # Set gene identifier as index
    # --------------------------------------------------------

    df = df.set_index(
        "Hugo_Symbol"
    )

    # --------------------------------------------------------
    # Convert patient columns to numeric
    # --------------------------------------------------------

    df = df.apply(
        pd.to_numeric,
        errors="coerce"
    )

    # Sort genes and patients
    df = df.sort_index()

    df = df.reindex(
        sorted(df.columns),
        axis=1
    )

    return df
````

````python
# ============================================================
# CELL 10.2 — ALIGN MOLECULAR MATRICES TO FINAL COHORT
# ============================================================

def align_matrix_to_patients(matrix, patient_ids):
    matrix = matrix.copy()

    matrix.columns = (
        matrix.columns
        .astype(str)
        .str.strip()
    )

    # Reindex preserves missing patients as NaN
    matrix = matrix.reindex(
        columns=patient_ids
    )

    return matrix


# ------------------------------------------------------------
# TCGA
# ------------------------------------------------------------

tcga_mut_aligned = align_matrix_to_patients(
    tcga_mutation_binary,
    tcga_patients_final
)

tcga_cna_aligned = align_matrix_to_patients(
    tcga_cna_matrix,
    tcga_patients_final
)

tcga_meth_aligned = align_matrix_to_patients(
    tcga_meth_matrix,
    tcga_patients_final
)


# ------------------------------------------------------------
# METABRIC
# ------------------------------------------------------------

metabric_mut_aligned = align_matrix_to_patients(
    metabric_mutation_binary,
    metabric_patients_final
)

metabric_cna_aligned = align_matrix_to_patients(
    metabric_cna_matrix,
    metabric_patients_final
)

metabric_meth_aligned = align_matrix_to_patients(
    metabric_meth_matrix,
    metabric_patients_final
)


print("=" * 80)
print("ALIGNED MOLECULAR MATRICES")
print("=" * 80)

print("\nTCGA")
print("Mutation    :", tcga_mut_aligned.shape)
print("CNA         :", tcga_cna_aligned.shape)
print("Methylation :", tcga_meth_aligned.shape)

print("\nMETABRIC")
print("Mutation    :", metabric_mut_aligned.shape)
print("CNA         :", metabric_cna_aligned.shape)
print("Methylation :", metabric_meth_aligned.shape)
````

### Cell 11.7 — 10-NN imputation

### Important technical point

The paper specifies **10-nearest-neighbor imputation using R's `impute.knn()`**.

Because you're implementing this in **Python**, we'll use scikit-learn's `KNNImputer` as the Python equivalent.

We transpose:

```text
features × patients
        ↓
patients × features
```

for the sklearn operation, then transpose back.

````python
# ============================================================
# CELL 11.7 : 10-NN IMPUTATION
# ============================================================


KNN_NEIGHBORS = 10

def perform_knn_imputation(
    matrix,
    n_neighbors=10
):
    """
    Perform 10-nearest-neighbor imputation.

    Input:
        features × patients

    KNNImputer:
        samples × features

    Therefore the matrix is transposed before
    imputation and restored afterward.
    """

    missing_count = int(
        matrix.isna()
        .sum()
        .sum()
    )

    if missing_count == 0:

        print(
            "No missing values detected. "
            "Skipping KNN imputation."
        )

        return matrix.copy()

    print(
        f"Imputing {missing_count:,} "
        f"missing values using "
        f"{n_neighbors}-NN..."
    )

    # Features × patients
    matrix_t = matrix.T

    # Patients × features
    imputer = KNNImputer(
        n_neighbors=n_neighbors
    )

    imputed_array = (
        imputer
        .fit_transform(matrix_t)
    )

    # Restore patients × features
    imputed_t = pd.DataFrame(
        imputed_array,
        index=matrix_t.index,
        columns=matrix_t.columns
    )

    # Restore features × patients
    imputed_matrix = imputed_t.T

    return imputed_matrix
````

````python
# ============================================================
# CELL 11.8 : APPLY 10-NN IMPUTATION
# ============================================================

dna_imputed = perform_knn_imputation(
    dna_filtered,
    n_neighbors=KNN_NEIGHBORS
)


print("=" * 80)
print("10-NN IMPUTATION COMPLETED")
print("=" * 80)

print(
    "Features :",
    dna_imputed.shape[0]
)

print(
    "Patients :",
    dna_imputed.shape[1]
)
````
